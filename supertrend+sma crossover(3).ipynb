```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
from itertools import product
import pandas_ta as ta
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from sklearn.preprocessing import MinMaxScaler
```

```python


class Strategy:
    def __init__(self, dataset, short, long, adx, tc=0.0005, amount=10, stop_multiplier=1.75, tp_multiplier=2.0, leverage=1):
        if not isinstance(short, int) or not isinstance(long, int) or short >= long or short <= 0:
            raise ValueError("Short must be a positive integer less than Long")
        self.data = dataset
        self.short = short
        self.long = long
        self.tc = tc
        self.amount = amount
        self.df = None
        self.results = None
        self.original_df = None
        self.adx = adx
        self.max_drawdown = 0
        self.avg_drawdown = 0
        self.annualized_sharpe = np.nan
        self.stop_multiplier = stop_multiplier  # Multiplier for ATR-based trailing stop
        self.tp_multiplier = tp_multiplier
        self.leverage = leverage 


    def preparedata(self):
        """Load and prepare the dataset from a CSV file."""
        self.df = pd.read_csv(self.data, index_col='Datetime', parse_dates=True)
        self.original_df = self.df.copy()

    def supertrend(self, df, periods=14, multiplier=3.0):
        """Calculate Supertrend and generate buy/sell signals."""
        st = ta.supertrend(
            high=self.df['High'],
            low=self.df['Low'],
            close=self.df['Close'],
            length=periods,
            multiplier=multiplier
        )
        supertrend_col = f"SUPERT_{periods}_{multiplier:.1f}"
        direction_col = f"SUPERTd_{periods}_{multiplier:.1f}"
        self.df['super_buy'] = (st[direction_col] > 0) & (st[direction_col].shift(1) < 0)
        self.df['super_sell'] = (st[direction_col] < 0) & (st[direction_col].shift(1) > 0)

    def sma_strategy(self):
        """Calculate SMAs, MACD, ATR, and generate buy/sell signals with 2 SMAs."""
        # Calculate Simple Moving Averages
        self.df['smashort'] = ta.sma(self.df['Close'], length=self.short)
        self.df['smalong'] = ta.sma(self.df['Close'], length=self.long)

        # Calculate ATR (length=14)
        self.df['atr'] = ta.atr(self.df['High'], self.df['Low'], self.df['Close'], length=14)

        # Define buy and sell signals based on 2 SMAs
        self.df['buy'] = self.df['smashort'] > self.df['smalong']
        self.df['sell'] = self.df['smashort'] < self.df['smalong']

    def ADX(self):
        """Calculate ADX and clean the dataframe."""
        adx = ta.adx(self.df['High'], self.df['Low'], self.df['Close'], length=20)
        self.df['adx'] = adx['ADX_20']
        self.df.dropna(subset=['smashort', 'smalong', 'atr', 'adx'], inplace=True)

    def execute_strategy(self):
        self.df['position'] = 0
        self.df['returns'] = self.df['Close'].pct_change().fillna(0)

        tp_level = None
        highest_price = None
        lowest_price = None
        stop_loss = None

        for i in range(1, len(self.df)):
            prev_pos = self.df['position'].iloc[i-1]
            close_price = self.df['Close'].iloc[i]
            atr = self.df['atr'].iloc[i]

            if prev_pos == 1:  # Long position
                if close_price >= tp_level:
                    self.df.at[self.df.index[i], 'position'] = 0
                    tp_level = None
                    highest_price = None
                    stop_loss = None
                else:
                    highest_price = max(highest_price, close_price) if highest_price else close_price
                    stop_loss = highest_price - (self.stop_multiplier * atr)
                    tp_level = highest_price + (self.tp_multiplier * atr)
                    if close_price <= stop_loss:
                        self.df.at[self.df.index[i], 'position'] = 0
                        tp_level = None
                        highest_price = None
                        stop_loss = None
                    else:
                        self.df.at[self.df.index[i], 'position'] = 1

            elif prev_pos == -1:  # Short position
                if close_price <= tp_level:
                    self.df.at[self.df.index[i], 'position'] = 0
                    tp_level = None
                    lowest_price = None
                    stop_loss = None
                else:
                    lowest_price = min(lowest_price, close_price) if lowest_price else close_price
                    stop_loss = lowest_price + (self.stop_multiplier * atr)
                    tp_level = lowest_price - (self.tp_multiplier * atr)
                    if close_price >= stop_loss:
                        self.df.at[self.df.index[i], 'position'] = 0
                        tp_level = None
                        lowest_price = None
                        stop_loss = None
                    else:
                        self.df.at[self.df.index[i], 'position'] = -1


            elif prev_pos == 0:  # No position, check for entry
                if (self.df['buy'].iloc[i]  and
                    self.df['adx'].iloc[i] > self.adx and
                    self.df['super_buy'].iloc[i]):
                    self.df.at[self.df.index[i], 'position'] = 1
                    tp_level = close_price + (atr * self.tp_multiplier)
                    highest_price = close_price
                    stop_loss = close_price - (atr * self.stop_multiplier)

                elif (self.df['sell'].iloc[i]  and
                      self.df['adx'].iloc[i] > self.adx and
                      self.df['super_sell'].iloc[i]):
                    self.df.at[self.df.index[i], 'position'] = -1
                    tp_level = close_price - (atr * self.tp_multiplier)
                    lowest_price = close_price
                    stop_loss = close_price + (atr * self.stop_multiplier)

                else:
                    self.df.at[self.df.index[i], 'position'] = 0

    def result(self):
        """Calculate strategy performance metrics."""
        self.results = self.df.copy()
        self.results['strategy'] = self.results['position'].shift(1).fillna(0) * self.results['returns'] * self.leverage
        self.results['trades'] = self.results['position'].diff().fillna(0).abs()
        self.results['strategy'] = self.results['strategy'] - (self.results['trades'] * self.tc)
        self.results['creturns'] = self.amount * (1 + self.results['returns']).cumprod()
        self.results['cstrategy'] = self.amount * (1 + self.results['strategy']).cumprod()

        # Calculate trade drawdowns
        current_position = 0
        trades = []
        current_trade = None

        for i in range(len(self.results)):
            new_position = self.results['position'].iloc[i]
            if current_position == 0 and new_position != 0:
                current_trade = {'entry_index': i, 'direction': new_position, 'exit_index': None, 'start_returns': i + 1}
            elif current_position != 0 and (new_position != current_position or new_position == 0):
                if current_trade is not None:
                    current_trade['exit_index'] = i
                    current_trade['end_returns'] = i
                    trades.append(current_trade)
                    current_trade = None
                if new_position != 0:
                    current_trade = {'entry_index': i, 'direction': new_position, 'exit_index': None, 'start_returns': i + 1}
            current_position = new_position

        if current_trade is not None:
            current_trade['exit_index'] = len(self.results) - 1
            current_trade['end_returns'] = len(self.results) - 1
            trades.append(current_trade)

        self.trade_drawdowns = []
        for trade in trades:
            start = trade['start_returns']
            end = trade['end_returns']
            if start > end or start >= len(self.results) or end >= len(self.results):
                continue
            trade_returns = self.results['strategy'].iloc[start:end+1]
            if len(trade_returns) == 0:
                continue
            cum_returns = (1 + trade_returns).cumprod()
            peak = cum_returns.cummax()
            drawdown = (cum_returns / peak) - 1
            max_dd = drawdown.min()
            self.trade_drawdowns.append(max_dd)

        if self.trade_drawdowns:
            self.max_drawdown = min(self.trade_drawdowns)
            self.avg_drawdown = sum(self.trade_drawdowns) / len(self.trade_drawdowns)
        else:
            self.max_drawdown = 0
            self.avg_drawdown = 0

        # Calculate annualized Sharpe ratio
        daily_cstrategy = self.results['cstrategy'].resample('D').last().dropna()
        if len(daily_cstrategy) > 1:
            daily_returns = daily_cstrategy.pct_change().dropna()
            if len(daily_returns) > 0 and daily_returns.std() > 0:
                self.annualized_sharpe = (daily_returns.mean() / daily_returns.std()) * np.sqrt(365)
            else:
                self.annualized_sharpe = np.nan
        else:
            self.annualized_sharpe = np.nan

    def plot_result(self):
        """Plot the strategy results using Plotly."""
        if self.results is None:
            print("Run result() first.")
            return

        try:
            # Create subplots (2 rows for price & returns, 1 row for ADX)
            fig = make_subplots(rows=3, cols=1, shared_xaxes=True, subplot_titles=("Price & Moving Averages", "Portfolio Performance", "ADX Indicator"))

            # --- Price & Moving Averages ---
            fig.add_trace(go.Scatter(x=self.results.index, y=self.results['Close'], mode='lines', name='Close'), row=1, col=1)
            fig.add_trace(go.Scatter(x=self.results.index, y=self.results['smashort'], mode='lines', name=f'SMA {self.short}'), row=1, col=1)
            fig.add_trace(go.Scatter(x=self.results.index, y=self.results['smalong'], mode='lines', name=f'SMA {self.long}'), row=1, col=1)
            
            # --- Portfolio Performance ---
            fig.add_trace(go.Scatter(x=self.results.index, y=self.results['creturns'].dropna(), mode='lines', name='Buy & Hold'), row=2, col=1)
            fig.add_trace(go.Scatter(x=self.results.index, y=self.results['cstrategy'].dropna(), mode='lines', name='Strategy'), row=2, col=1)

            # --- ADX Indicator ---
            fig.add_trace(go.Scatter(x=self.results.index, y=self.results['adx'], mode='lines', name='ADX', line=dict(color='purple', dash='dash')), row=3, col=1)
            fig.add_hline(y=20, line=dict(color='red', dash='dash', width=0.7), annotation_text="Level 20", row=3, col=1)
            fig.add_hline(y=25, line=dict(color='orange', dash='dash', width=0.7), annotation_text="Level 25", row=3, col=1)
            fig.add_hline(y=30, line=dict(color='green', dash='dash', width=0.7), annotation_text="Level 30", row=3, col=1)

            # Layout Formatting
            fig.update_layout(
                title=f"SMA Strategy | Short = {self.short} Long = {self.long} | TC = {self.tc}",
                height=900,
                width=1200,
                legend=dict(x=0.02, y=1.02),
                template="plotly_dark"
            )

            fig.show()

            print(f"Total Returns: {self.results['cstrategy'].iloc[-1]:.2f}")
            print(f"Total Trades: {int(self.results['trades'].sum())}")
            print(f"Trade Frequency: {self.results['trades'].sum() / len(self.results):.4f}")
            print(f"Maximum Drawdown per Trade: {self.max_drawdown:.2%}")
            print(f"Average Drawdown per Trade: {self.avg_drawdown:.2%}")
            if not np.isnan(self.annualized_sharpe):
                print(f"Annualized Sharpe Ratio: {self.annualized_sharpe:.2f}")
            else:
                print("Annualized Sharpe Ratio: N/A")
        except Exception as e:
            print(f"Error in plot_result: {str(e)}")

    def optimize_strategy(self, shortrange, longrange, adx_range):
        """Optimize strategy parameters based on a combination of return, trades, and Sharpe ratio."""
        combinations = list(product(shortrange, longrange, adx_range))
        perf = []
        
        for short, long, adx in combinations:
            if short >= long or short <= 0:
                continue
            
            self.df = self.original_df.copy()
            self.short = short
            self.long = long
            self.adx = adx
            min_rows_needed = max(self.long, 26)
            if len(self.df) < min_rows_needed:
                continue

            self.sma_strategy()
            self.ADX()
            self.supertrend(self.df)
            self.execute_strategy()
            self.result()
            
            if self.results is None:
                continue
            
            total_return = self.results['cstrategy'].iloc[-1]
            total_trades = self.results['trades'].sum()
            # Replace NaN Sharpe with a large negative value
            sharpe_ratio = self.annualized_sharpe if not np.isnan(self.annualized_sharpe) else -1e10

            perf.append((short, long, adx, total_return, total_trades, sharpe_ratio))
        
        if not perf:
            raise ValueError("No valid combinations found")
        
        # Convert to NumPy array and replace any NaN or infinite values with finite numbers
        perf_array = np.array(perf, dtype=np.float64)
        perf_array = np.nan_to_num(perf_array, nan=0.0, posinf=1e10, neginf=-1e10)
        
        # Normalize metrics using MinMax scaling (columns: total_return, total_trades, Sharpe)
        scaler = MinMaxScaler()
        normalized_metrics = scaler.fit_transform(perf_array[:, 3:])
        
        # Assign weights (40% return, 30% trades, 30% Sharpe)
        weights = np.array([0.4, 0.3, 0.3])
        scores = np.dot(normalized_metrics, weights)
        
        best_idx = np.argmax(scores)
        best = perf[best_idx]

        # Apply best parameters
        self.short, self.long, self.adx = best[0], best[1], best[2]
        self.df = self.original_df.copy()
        self.sma_strategy()
        self.ADX()
        self.supertrend(self.df)
        self.execute_strategy()
        self.result()
        self.plot_result()

        print(f"Best Combo: Short = {self.short}, Long = {self.long}, ADX = {self.adx}")
        print(f"Total Returns: {best[3]:.2f}")
        print(f"Total Trades: {int(best[4])}")
        print(f"Trade Frequency: {best[4] / len(self.results):.4f}")
        print(f"Sharpe Ratio: {best[5]:.2f}")


# Example usage
try:
    data = Strategy('ETH1year15min.csv', short=19, long=32, adx=16, leverage=5)
    data.preparedata()
    data.sma_strategy()
    data.ADX()
    data.supertrend(data.df)
    data.execute_strategy()
    data.result()
    data.plot_result()

    # Uncomment to optimize parameters
    """data.optimize_strategy(
         shortrange=range(10, 21, 3),  # e.g., 10, 13, 16, 19
         longrange=range(25, 51, 3),   # e.g., 32, 35, 38, ..., 50
         adx_range=range(15, 41, 3)    # e.g., 20, 23, 26, ..., 40
     )"""
except Exception as e:
    print(f"Main execution error: {str(e)}")
```
