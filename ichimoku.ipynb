```python
import pandas as pd
import numpy as np
import plotly
import pandas_ta as ta
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python

class Ichimoku:
    def __init__(self, data: str, tp_multiplier: float = 2, sp_multiplier: float = 1.5,
                 leverage: int = 1, tc: float = 0.0005, amount: int = 10) -> None:
        """Initialize Ichimoku strategy parameters."""
        self.tp_multiplier = tp_multiplier
        self.sp_multiplier = sp_multiplier
        self.data = data
        self.leverage = leverage
        self.tc = tc  # Transaction cost
        self.amount = amount
        self.df = pd.DataFrame()
        self.df_ichi = pd.DataFrame()
        self.results = None
        self.trade_drawdowns = []
        self.max_drawdown = 0
        self.avg_drawdown = 0
        self.annualized_sharpe = np.nan

    def prepdata(self) -> None:
        self.df = pd.read_csv(self.data, parse_dates=['Datetime'])
        self.df.set_index('Datetime', inplace=True)
        
        ichimoku_data = ta.ichimoku(self.df['High'], self.df['Low'], self.df['Close'])
        self.df_ichi = ichimoku_data[0].rename(columns={
            'ISA_9': 'leading Span A',
            'ISB_26': 'leading Span B',
            'ITS_9': 'conversion line',
            'IKS_26': 'base line',
            'ICS_26': 'lagging Span'
        })
        
        self.df_ichi.index = self.df.index  # Align both dataframes


    def strategy(self) -> None:
        
        """Define Ichimoku buy/sell conditions and calculate ATR."""
        # Buy conditions
        self.df['buy_cond1'] = self.df_ichi['lagging Span'] > self.df['Close'].shift(26)
        self.df['buy_cond2'] = self.df_ichi['conversion line'] > self.df_ichi['base line']
        self.df['buy_cond3'] = self.df_ichi['leading Span A'] > self.df_ichi['leading Span B']
        self.df['buy_cond4'] = self.df['Close'] > self.df_ichi['leading Span A']

        # Sell conditions
        self.df['sell_cond1'] = self.df_ichi['lagging Span'] < self.df['Close'].shift(26)
        self.df['sell_cond2'] = self.df_ichi['conversion line'] < self.df_ichi['base line']
        self.df['sell_cond3'] = self.df_ichi['leading Span A'] < self.df_ichi['leading Span B']
        self.df['sell_cond4'] = self.df['Close'] < self.df_ichi['leading Span A']

        self.df['atr'] = ta.atr(self.df['High'], self.df['Low'], self.df['Close'], length=14)

       

    

    def execute_trade(self) -> None:
        """Execute trades based on predefined conditions and manage positions."""
        self.df['position'] = 0
        self.df['returns'] = self.df['Close'].pct_change().fillna(0)

        tp_level = None
        highest_price = None
        lowest_price = None
        stop_loss = None

        for i in range(1, len(self.df)):
            prev_pos = self.df['position'].iloc[i-1]
            close_price = self.df['Close'].iloc[i]
            atr = self.df['atr'].iloc[i]

            if prev_pos == 1:  # Long position
                if close_price >= tp_level:
                    self.df.at[self.df.index[i], 'position'] = 0
                    tp_level = None
                    highest_price = None
                    stop_loss = None
                else:
                    highest_price = max(highest_price, close_price) if highest_price else close_price
                    stop_loss = highest_price - (self.sp_multiplier * atr)
                    tp_level = highest_price + (self.tp_multiplier * atr)
                    if close_price <= stop_loss:
                        self.df.at[self.df.index[i], 'position'] = 0
                        tp_level = None
                        highest_price = None
                        stop_loss = None
                    else:
                        self.df.at[self.df.index[i], 'position'] = 1

            elif prev_pos == -1:  # Short position
                if close_price <= tp_level:
                    self.df.at[self.df.index[i], 'position'] = 0
                    tp_level = None
                    lowest_price = None
                    stop_loss = None
                else:
                    lowest_price = min(lowest_price, close_price) if lowest_price else close_price
                    stop_loss = lowest_price + (self.sp_multiplier * atr)
                    tp_level = lowest_price - (self.tp_multiplier * atr)
                    if close_price >= stop_loss:
                        self.df.at[self.df.index[i], 'position'] = 0
                        tp_level = None
                        lowest_price = None
                        stop_loss = None
                    else:
                        self.df.at[self.df.index[i], 'position'] = -1

            elif prev_pos == 0:  # No position
                if (self.df['buy_cond1'].iloc[i] and
                    self.df['buy_cond2'].iloc[i] and
                     self.df['buy_cond3'].iloc[i] and
                      self.df['buy_cond4'].iloc[i] ):
                    self.df.at[self.df.index[i], 'position'] = 1
                    tp_level = close_price + (atr * self.tp_multiplier)
                    highest_price = close_price
                    stop_loss = close_price - (atr * self.sp_multiplier)

                elif (self.df['sell_cond1'].iloc[i] and
                    self.df['sell_cond2'].iloc[i] and
                     self.df['sell_cond3'].iloc[i] and
                      self.df['sell_cond4'].iloc[i] ):
                    self.df.at[self.df.index[i], 'position'] = -1
                    tp_level = close_price - (atr * self.tp_multiplier)
                    lowest_price = close_price
                    stop_loss = close_price + (atr * self.sp_multiplier)

    def result(self) -> None:
        """Calculate performance metrics and trade statistics."""
        self.results = self.df.copy()
        self.results['strategy'] = self.results['position'].shift(1).fillna(0) * self.results['returns'] * self.leverage
        self.df['trades'] = self.results['position'].diff().fillna(0).abs().cumsum()
        self.results['trades'] = self.results['position'].diff().fillna(0).abs()
        self.results['strategy'] = self.results['strategy'] - (self.results['trades'] *self.tc)
        self.results['creturns'] = self.amount * (1 + self.results['returns']).cumprod()
        self.results['cstrategy'] = self.amount * (1 + self.results['strategy']).cumprod()

        current_position = 0
        trades = []
        current_trade = None

        for i in range(len(self.results)):
            new_position = self.results['position'].iloc[i]
            if current_position == 0 and new_position != 0:
                current_trade = {'entry_index': i, 'direction': new_position, 'exit_index': None, 'start_returns': i + 1}
            elif current_position != 0 and new_position != current_position:
                if current_trade is not None:
                    current_trade['exit_index'] = i
                    current_trade['end_returns'] = i
                    trades.append(current_trade)
                    current_trade = None
                if new_position != 0:
                    current_trade = {'entry_index': i, 'direction': new_position, 'exit_index': None, 'start_returns': i + 1}
            current_position = new_position

        if current_trade is not None:
            current_trade['exit_index'] = len(self.results) - 1
            current_trade['end_returns'] = len(self.results) - 1
            trades.append(current_trade)

        self.trade_drawdowns = []
        for trade in trades:
            start = trade['start_returns']
            end = trade['end_returns']
            if start >= len(self.results) or end >= len(self.results):
                continue
            trade_returns = self.results['strategy'].iloc[start:end+1]
            if len(trade_returns) == 0:
                continue
            cum_returns = (1 + trade_returns).cumprod()
            peak = cum_returns.cummax()
            drawdown = (cum_returns / peak) - 1
            max_dd = drawdown.min()
            self.trade_drawdowns.append(max_dd)

        self.max_drawdown = min(self.trade_drawdowns) if self.trade_drawdowns else 0
        self.avg_drawdown = sum(self.trade_drawdowns)/len(self.trade_drawdowns) if self.trade_drawdowns else 0

        # Annualized Sharpe Ratio
        daily_returns = self.results['cstrategy'].resample('D').last().pct_change().dropna()
        if not daily_returns.empty and daily_returns.std() != 0:
            self.annualized_sharpe = (daily_returns.mean() / daily_returns.std()) * np.sqrt(365)
        else:
            self.annualized_sharpe = np.nan

    def plot_result(self) -> None:
        """Visualize the strategy performance and indicators using Plotly."""
        if self.results is None:
            print("Execute result() method first.")
            return

        fig = make_subplots(rows=2, cols=1, shared_xaxes=True, subplot_titles=("Price with Ichimoku", "Cumulative Returns"))

        # Price and Ichimoku
        fig.add_trace(go.Candlestick(x=self.df_ichi.index, open=self.df['Open'], high=self.df['High'],
                                    low=self.df['Low'], close=self.df['Close'], name='Price'), row=1, col=1)
        fig.add_trace(go.Scatter(x=self.df_ichi.index, y=self.df_ichi['leading Span A'], 
                              line=dict(color='green'), name='Leading Span A'), row=1, col=1)
        fig.add_trace(go.Scatter(x=self.df_ichi.index, y=self.df_ichi['leading Span B'], 
                              line=dict(color='red'), name='Leading Span B'), row=1, col=1)
        fig.add_trace(go.Scatter(x=self.df_ichi.index, y=self.df_ichi['conversion line'], 
                              line=dict(color='blue'), name='Conversion Line'), row=1, col=1)
        fig.add_trace(go.Scatter(x=self.df_ichi.index, y=self.df_ichi['base line'], 
                              line=dict(color='purple'), name='Base Line'), row=1, col=1)
        fig.add_trace(go.Scatter(x=self.df_ichi.index, y=self.df_ichi['lagging Span'], 
                              line=dict(color='orange'), name='Lagging Span'), row=1, col=1)

        # Cumulative Returns
        fig.add_trace(go.Scatter(x=self.results.index, y=self.results['creturns'], 
                               name='Buy & Hold', line=dict(color='gray')), row=2, col=1)
        fig.add_trace(go.Scatter(x=self.results.index, y=self.results['cstrategy'], 
                               name='Strategy', line=dict(color='cyan')), row=2, col=1)

        fig.update_layout(height=800, width=1200, title_text="Ichimoku Strategy Analysis", template="plotly_dark")
        fig.show()

        # Performance Summary
        
        print(f"Final Portfolio Value: ${self.results['cstrategy'].iloc[-1]:.2f}")
        print(f"Maximum Drawdown: {self.max_drawdown:.2%}")
        print(f"Sharpe Ratio: {self.annualized_sharpe:.2f}")
        print(self.df[['Close', 'position', 'trades']].tail(30))

        

# Example Execution
if __name__ == "__main__":
    try:
        strategy = Ichimoku('ETH1year15min.csv')
        strategy.prepdata()
        strategy.strategy()
        strategy.execute_trade()
        strategy.result()
        strategy.plot_result()
    except Exception as e:
        print(f"Execution halted due to: {str(e)}")
```
