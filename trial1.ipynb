```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

def pivot_High(prices, left, right):
    """Find pivot Highs in a series."""
    pivots = np.full(len(prices), np.nan)
    for i in range(left, len(prices) - right):
        if prices[i] == max(prices[i - left:i + right + 1]):
            pivots[i] = prices[i]
    return pivots

def pivot_Low(prices, left, right):
    """Find pivot Lows in a series."""
    pivots = np.full(len(prices), np.nan)
    for i in range(left, len(prices) - right):
        if prices[i] == min(prices[i - left:i + right + 1]):
            pivots[i] = prices[i]
    return pivots

def value_when(condition, values, occurrence):
    """Return the value at the nth occurrence of a condition."""
    filtered_values = values[condition]
    if len(filtered_values) > occurrence:
        return filtered_values.iloc[-(occurrence + 1)]
    return np.nan

def calculate_levels(df, left=50, right=25, quick_right=5, source='Close'):
    High = df['High']
    Low = df['Low']
    Close = df['Close']
    
    if source == 'Close':
        pivot_Highs = pivot_High(Close, left, right)
        pivot_Lows = pivot_Low(Close, left, right)
        quick_pivot_Highs = pivot_High(Close, left, quick_right)
        quick_pivot_Lows = pivot_Low(Close, left, quick_right)
    else:
        pivot_Highs = pivot_High(High, left, right)
        pivot_Lows = pivot_Low(Low, left, right)
        quick_pivot_Highs = pivot_High(High, left, quick_right)
        quick_pivot_Lows = pivot_Low(Low, left, quick_right)
    
    levels = {}
    levels['level1'] = value_when(pd.Series(quick_pivot_Highs).notna(), Close.shift(quick_right), 0)
    levels['level2'] = value_when(pd.Series(quick_pivot_Lows).notna(), Close.shift(quick_right), 0)
    levels['level3'] = value_when(pd.Series(pivot_Highs).notna(), Close.shift(right), 0)
    levels['level4'] = value_when(pd.Series(pivot_Lows).notna(), Close.shift(right), 0)
    levels['level5'] = value_when(pd.Series(pivot_Highs).notna(), Close.shift(right), 1)
    levels['level6'] = value_when(pd.Series(pivot_Lows).notna(), Close.shift(right), 1)
    levels['level7'] = value_when(pd.Series(pivot_Highs).notna(), Close.shift(right), 2)
    levels['level8'] = value_when(pd.Series(pivot_Lows).notna(), Close.shift(right), 2)
    
    return levels

# Example usage with a DataFrame containing OHLC data
df = pd.read_csv('bitcoin1year15min.csv')

levels = calculate_levels(df)
print(levels)
def plot_levels(df, levels):
    fig = go.Figure()

    # Add candlestick chart
    fig.add_trace(go.Candlestick(x=df['Datetime'],
                                 Open=df['Open'],
                                 High=df['High'],
                                 Low=df['Low'],
                                 Close=df['Close'],
                                 name='Candlestick'))

    # Add horizontal lines for levels
    for level_name, level_value in levels.items():
        if not pd.isna(level_value):
            fig.add_trace(go.Scatter(x=df['Datetime'],
                                     y=[level_value] * len(df),
                                     mode='lines',
                                     name=level_name,
                                     line=dict(color='blue', width=2)))

    fig.update_layout(title='Candlestick Chart with Support and Resistance Levels',
                      xaxis_title='Datetime',
                      yaxis_title='Price',
                      xaxis_rangeslider_visible=False)

    fig.show()

plot_levels(df, levels)
```

```python
import pandas as pd
from plotly.subplots import make_subplots
import matplotlib.pyplot as plt
import plotly.graph_objects as go

# Sample data
# Replace this with your actual data
# data = pd.read_csv('your_data.csv')
data = pd.read_csv('ETHUSDT3Month5min.csv')


left = 50
right = 25
quick_right = 5
src = 'Close'

# Pivot High and Low calculations

def pivotHigh(data, left, right):
    return data['High'].rolling(window=left + right + 1, center=True).max()

def pivotLow(data, left, right):
    return data['Low'].rolling(window=left + right + 1, center=True).min()

pivot_High = pivotHigh(data, left, right) if src == 'Close' else pivotHigh(data, left, right)
pivot_Lows = pivotLow(data, left, right) if src == 'Close' else pivotLow(data, left, right)

quick_pivot_High = pivotHigh(data, left, quick_right) if src == 'Close' else pivotHigh(data, left, quick_right)
quick_pivot_Lows = pivotLow(data, left, quick_right) if src == 'Close' else pivotLow(data, left, quick_right)

# Level calculations
level1 = quick_pivot_High.shift(quick_right) if src == 'Close' else quick_pivot_High.shift(quick_right)
level2 = quick_pivot_Lows.shift(quick_right) if src == 'Close' else quick_pivot_Lows.shift(quick_right)
level3 = pivot_High.shift(right) if src == 'Close' else pivot_High.shift(right)
level4 = pivot_Lows.shift(right) if src == 'Close' else pivot_Lows.shift(right)
level5 = pivot_High.shift(right + 1) if src == 'Close' else pivot_High.shift(right + 1)
level6 = pivot_Lows.shift(right + 1) if src == 'Close' else pivot_Lows.shift(right + 1)
level7 = pivot_High.shift(right + 2) if src == 'Close' else pivot_High.shift(right + 2)
level8 = pivot_Lows.shift(right + 2) if src == 'Close' else pivot_Lows.shift(right + 2)

# Plotting
# Ensure levels are not NaN before plotting
levels = {
    'level1': level1,
    'level2': level2,
    'level3': level3,
    'level4': level4,
    'level5': level5,
    'level6': level6,
    'level7': level7,
    'level8': level8
}

# levels_to_plot = {k: v for k, v in levels.items() if not pd.isna(v).all()}
levels_to_plot = levels

# Filter out NaN values from each level
filtered_levels = {}
for name, level in levels_to_plot.items():
    filtered_levels[name] = level.dropna()

# Get the last 8 levels
last_8_levels = dict(list(filtered_levels.items())[-8:])

# Create subplots
fig = make_subplots(rows=1, cols=1, shared_xaxes=True)

# Add candlestick trace
fig.add_trace(go.Candlestick(x=data['Open Time'],
                             Open=data['Open'],
                             High=data['High'],
                             Low=data['Low'],
                             Close=data['Close'],
                             name='Candlestick'),
              row=1, col=1)

# Add horizontal lines for support and resistance levels
for name, level in last_8_levels.items():
    # Use only the last known value for each level
    last_level_value = level.iloc[-1] if not level.empty else None
    
    if last_level_value is not None:
        fig.add_trace(go.Scatter(x=data['Open Time'],
                                 y=[last_level_value] * len(data),
                                 mode='lines',
                                 name=name,
                                 line=dict(color='blue', width=2)),
                      row=1, col=1)

fig.update_layout(title='Candlestick Chart with Support and Resistance Levels (Last 8)',
                  xaxis_title='Datetime',
                  yaxis_title='Price',
                  xaxis_rangeslider_visible=False)

fig.show()
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

def calculate_market_structure(
    df, 
    len_period=50, 
    short_len=3,
    bullCss='#089981', 
    bearCss='#ff5252',
    showChoch=True, 
    showBos=True,
    showIdm=True, 
    idmCss='gray',
    showSweeps=True, 
    sweepsCss='gray',
    showCircles=True
):
    """
    Replicates the logic of the Pine Script indicator for Market Structure with Inducements & Sweeps.
    Assumes that the DataFrame 'df' has columns: 'Open', 'High', 'Low', 'Close', and a datetime index.
    """
    # Initialize state variables
    os_state = 0
    top_crossed = False
    btm_crossed = False
    max_val = np.nan
    min_val = np.nan
    max_x1 = np.nan
    min_x1 = np.nan
    topy = np.nan
    btmy = np.nan
    stop_crossed = False
    sbtm_crossed = False

    n_bars = df.shape[0]

    # Preallocate arrays for swing detection
    swing_top   = [np.nan] * n_bars
    swing_topx  = [np.nan] * n_bars
    swing_btm   = [np.nan] * n_bars
    swing_btmx  = [np.nan] * n_bars

    swing_stop  = [np.nan] * n_bars
    swing_stopx = [np.nan] * n_bars
    swing_sbtm  = [np.nan] * n_bars
    swing_sbtmx = [np.nan] * n_bars

    # --- Primary Swing Detection (period = len_period) ---
    os_primary = 0
    for i in range(len_period, n_bars):
        window = df.iloc[i-len_period+1 : i+1]
        upper = window['High'].max()
        lower = window['Low'].min()
        ref_idx = i - len_period
        if ref_idx < 0:
            continue
        # Compare bar at ref_idx to window extremes
        if df.iloc[ref_idx]['High'] > upper:
            new_os = 0
        elif df.iloc[ref_idx]['Low'] < lower:
            new_os = 1
        else:
            new_os = os_primary

        # Detect new swing
        if new_os != os_primary:
            if new_os == 0:
                # Swing high
                swing_top[i]  = df.iloc[ref_idx]['High']
                swing_topx[i] = ref_idx
            elif new_os == 1:
                # Swing low
                swing_btm[i]  = df.iloc[ref_idx]['Low']
                swing_btmx[i] = ref_idx

        os_primary = new_os

    # --- Short Swing Detection (period = short_len) for IDM ---
    os_short = 0
    for i in range(short_len, n_bars):
        window = df.iloc[i-short_len+1 : i+1]
        upper = window['High'].max()
        lower = window['Low'].min()
        ref_idx = i - short_len
        if ref_idx < 0:
            continue

        if df.iloc[ref_idx]['High'] > upper:
            new_os = 0
        elif df.iloc[ref_idx]['Low'] < lower:
            new_os = 1
        else:
            new_os = os_short

        if new_os != os_short:
            if new_os == 0:
                swing_stop[i]  = df.iloc[ref_idx]['High']
                swing_stopx[i] = ref_idx
            elif new_os == 1:
                swing_sbtm[i]  = df.iloc[ref_idx]['Low']
                swing_sbtmx[i] = ref_idx

        os_short = new_os

    # Prepare containers for shapes, annotations, and swing circles
    shapes = []
    annotations = []
    circles_x = []
    circles_y = []
    circles_color = []

    # Helper function to get the actual x-value (date or index) for a bar number
    def x_val(bar_index):
        return df.index[bar_index]

    # Iterate over each bar
    for i in range(n_bars):
        # If we have a new swing high or low
        if not np.isnan(swing_top[i]):
            topy = swing_top[i]
            top_crossed = False
        if not np.isnan(swing_btm[i]):
            btmy = swing_btm[i]
            btm_crossed = False

        # --- CHoCH Detection ---
        # Bullish CHoCH
        if not np.isnan(topy) and df.iloc[i]['Close'] > topy and not top_crossed:
            os_state = 1
            top_crossed = True
            max_val = df.iloc[i]['High']
            min_val = df.iloc[i]['Low']
            max_x1 = i
            min_x1 = i
            stop_crossed = False
            sbtm_crossed = False

            if showChoch and not np.isnan(swing_topx[i]):
                shapes.append({
                    'type': 'line',
                    'xref': 'x', 'yref': 'y',
                    'x0': x_val(swing_topx[i]),
                    'y0': topy,
                    'x1': x_val(i),
                    'y1': topy,
                    'line': {'color': bullCss, 'dash': 'dash'}
                })
                annotations.append({
                    'x': x_val(i),
                    'y': topy,
                    'text': 'CHoCH',
                    'showarrow': False,
                    'font': {'color': bullCss, 'size': 10}
                })

        # Bearish CHoCH
        if not np.isnan(btmy) and df.iloc[i]['Close'] < btmy and not btm_crossed:
            os_state = 0
            btm_crossed = True
            max_val = df.iloc[i]['High']
            min_val = df.iloc[i]['Low']
            max_x1 = i
            min_x1 = i
            stop_crossed = False
            sbtm_crossed = False

            if showChoch and not np.isnan(swing_btmx[i]):
                shapes.append({
                    'type': 'line',
                    'xref': 'x', 'yref': 'y',
                    'x0': x_val(swing_btmx[i]),
                    'y0': btmy,
                    'x1': x_val(i),
                    'y1': btmy,
                    'line': {'color': bearCss, 'dash': 'dash'}
                })
                annotations.append({
                    'x': x_val(i),
                    'y': btmy,
                    'text': 'CHoCH',
                    'showarrow': False,
                    'font': {'color': bearCss, 'size': 10}
                })

        # --- Bullish BOS and IDM ---
        # Inducement (IDM)
        if (not np.isnan(swing_sbtm[i]) and 
            df.iloc[i]['Low'] < swing_sbtm[i] and 
            not sbtm_crossed and 
            os_state == 1 and 
            (np.isnan(btmy) or swing_sbtm[i] != btmy)):
            if showIdm and not np.isnan(swing_sbtmx[i]):
                shapes.append({
                    'type': 'line',
                    'xref': 'x', 'yref': 'y',
                    'x0': x_val(swing_sbtmx[i]),
                    'y0': swing_sbtm[i],
                    'x1': x_val(i),
                    'y1': swing_sbtm[i],
                    'line': {'color': idmCss, 'dash': 'dot'}
                })
                annotations.append({
                    'x': x_val(i),
                    'y': swing_sbtm[i],
                    'text': 'IDM',
                    'showarrow': False,
                    'font': {'color': idmCss, 'size': 10}
                })
            sbtm_crossed = True

        # BOS
        if (not np.isnan(max_val) and 
            df.iloc[i]['Close'] > max_val and 
            sbtm_crossed and os_state == 1):
            if showBos:
                shapes.append({
                    'type': 'line',
                    'xref': 'x', 'yref': 'y',
                    'x0': x_val(max_x1),
                    'y0': max_val,
                    'x1': x_val(i),
                    'y1': max_val,
                    'line': {'color': bullCss, 'dash': 'solid'}
                })
                annotations.append({
                    'x': x_val(i),
                    'y': max_val,
                    'text': 'BOS',
                    'showarrow': False,
                    'font': {'color': bullCss, 'size': 10}
                })
            sbtm_crossed = False

        # --- Bearish BOS and IDM ---
        # Inducement (IDM)
        if (not np.isnan(swing_stop[i]) and 
            df.iloc[i]['High'] > swing_stop[i] and 
            not stop_crossed and 
            os_state == 0 and 
            (np.isnan(topy) or swing_stop[i] != topy)):
            if showIdm and not np.isnan(swing_stopx[i]):
                shapes.append({
                    'type': 'line',
                    'xref': 'x', 'yref': 'y',
                    'x0': x_val(swing_stopx[i]),
                    'y0': swing_stop[i],
                    'x1': x_val(i),
                    'y1': swing_stop[i],
                    'line': {'color': 'gray', 'dash': 'dot'}
                })
                annotations.append({
                    'x': x_val(i),
                    'y': swing_stop[i],
                    'text': 'IDM',
                    'showarrow': False,
                    'font': {'color': 'gray', 'size': 10}
                })
            stop_crossed = True

        # BOS
        if (not np.isnan(min_val) and 
            df.iloc[i]['Close'] < min_val and 
            stop_crossed and os_state == 0):
            if showBos:
                shapes.append({
                    'type': 'line',
                    'xref': 'x', 'yref': 'y',
                    'x0': x_val(min_x1),
                    'y0': min_val,
                    'x1': x_val(i),
                    'y1': min_val,
                    'line': {'color': bearCss, 'dash': 'solid'}
                })
                annotations.append({
                    'x': x_val(i),
                    'y': min_val,
                    'text': 'BOS',
                    'showarrow': False,
                    'font': {'color': bearCss, 'size': 10}
                })
            stop_crossed = False

        # --- Sweeps Detection ---
        if (not np.isnan(max_val) and 
            df.iloc[i]['High'] > max_val and 
            df.iloc[i]['Close'] < max_val and 
            os_state == 1 and (i - max_x1 > 1) and showSweeps):
            shapes.append({
                'type': 'line',
                'xref': 'x', 'yref': 'y',
                'x0': x_val(max_x1),
                'y0': max_val,
                'x1': x_val(i),
                'y1': max_val,
                'line': {'color': sweepsCss, 'dash': 'dot'}
            })
            annotations.append({
                'x': x_val(i),
                'y': max_val,
                'text': 'x',
                'showarrow': False,
                'font': {'color': sweepsCss, 'size': 10}
            })

        if (not np.isnan(min_val) and 
            df.iloc[i]['Low'] < min_val and 
            df.iloc[i]['Close'] > min_val and 
            os_state == 0 and (i - min_x1 > 1) and showSweeps):
            shapes.append({
                'type': 'line',
                'xref': 'x', 'yref': 'y',
                'x0': x_val(min_x1),
                'y0': min_val,
                'x1': x_val(i),
                'y1': min_val,
                'line': {'color': sweepsCss, 'dash': 'dot'}
            })
            annotations.append({
                'x': x_val(i),
                'y': min_val,
                'text': 'x',
                'showarrow': False,
                'font': {'color': sweepsCss, 'size': 10}
            })

        # Update trailing max/min
        if np.isnan(max_val) or df.iloc[i]['High'] > max_val:
            max_val = df.iloc[i]['High']
            max_x1 = i
        if np.isnan(min_val) or df.iloc[i]['Low'] < min_val:
            min_val = df.iloc[i]['Low']
            min_x1 = i

        # --- Optional: Record swing circles ---
        if showCircles:
            if not np.isnan(swing_top[i]):
                circles_x.append(x_val(i))
                circles_y.append(swing_top[i])
                circles_color.append(bearCss)
            if not np.isnan(swing_btm[i]):
                circles_x.append(x_val(i))
                circles_y.append(swing_btm[i])
                circles_color.append(bullCss)

    # --- Create Plotly Figure ---
    fig = go.Figure()

    # Price Candlestick
    fig.add_trace(go.Candlestick(
        x=df.index,
        open=df['Open'],
        high=df['High'],
        low=df['Low'],
        close=df['Close'],
        name='Price'
    ))

    # Add shapes (lines)
    for shape in shapes:
        fig.add_shape(**shape)

    # Add annotations
    for ann in annotations:
        fig.add_annotation(**ann)

    # Add swing circles
    if showCircles and len(circles_x) > 0:
        fig.add_trace(go.Scatter(
            x=circles_x,
            y=circles_y,
            mode='markers',
            marker=dict(color=circles_color, size=8),
            name='Swings'
        ))

    # Final layout adjustments
    fig.update_layout(
        title="Market Structure with Inducements & Sweeps",
        xaxis_title="Date",
        yaxis_title="Price",
        xaxis_rangeslider_visible=False  # hide range slider if you want
    )

    return fig

# Example usage:
df = pd.read_csv('bitcoin_5_year.csv', parse_dates=True, index_col='Date')
fig = calculate_market_structure(df)
fig.show()
```
